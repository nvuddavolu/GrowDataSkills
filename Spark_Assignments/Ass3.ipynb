```python
spark = SparkSession.builder.appName('Assignment-3').getOrCreate()
```

```python
from pyspark.sql.types import *
File_schema = StructType([StructField('PatientID', StringType()), StructField('Age', IntegerType()), StructField('Gender',StringType()), StructField('DiagCode', StringType()), StructField('DiagDesc', StringType()), StructField('DiagDate', DateType())])

Raw_df = spark.read.format('csv').options(header = 'True').schema(File_schema).load('/Ass3/Input/*.csv')
```

```python
from pyspark.sql.functions import col, sum

Raw_df.agg(*[sum(col(column).isNull().cast(IntegerType())).alias(f'{column}_NullCount') for column in Raw_df.columns]).show()
```
